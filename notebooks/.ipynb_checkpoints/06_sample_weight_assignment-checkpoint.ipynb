### Use continuous weights instead of one fixed weight

Instead of assigning every outlier the same weight, we can make more extreme outliers receive smaller weights.

For example, suppose we have anomaly scores from IsolationForest:

```python
scores = iso.decision_function(X_train)
```

Usually, lower scores mean more abnormal observations.

We can convert scores into weights:

```python
import numpy as np

# Convert anomaly scores into percentile ranks
rank = np.argsort(np.argsort(scores)) / (len(scores) - 1)

# Lower rank = more abnormal
sample_weight = 0.3 + 0.7 * rank
```

This gives weights roughly between:
- Most abnormal records: weight near 0.3
- Most normal records:   weight near 1.0

We can make the floor more or less aggressive:

```python
min_weight = 0.5
sample_weight = min_weight + (1 - min_weight) * rank

# OR

min_weight = 0.2
sample_weight = min_weight + (1 - min_weight) * rank
```
