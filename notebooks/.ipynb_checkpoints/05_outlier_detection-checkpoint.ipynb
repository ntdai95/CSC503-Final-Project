# Load Preprocessed Data

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest


# File name
PREPROCESSED_DATA_FILE_NAME = "loan_data_preprocessed.csv"

PROJECT_ROOT = Path("..")
PREPROCESSED_DATA_FILE_PATH = PROJECT_ROOT / "data" / "pre-processed" / PREPROCESSED_DATA_FILE_NAME

OUTPUT_DIR = PROJECT_ROOT / "data" / "pre-processed"

print("PROJECT_ROOT:", PROJECT_ROOT)
print("Preprocessed data path:", PREPROCESSED_DATA_FILE_PATH)
print("Output folder:", OUTPUT_DIR)

df = pd.read_csv(PREPROCESSED_DATA_FILE_PATH)
print(df.shape)
df.head()
```

# Apply Isolation Forest on Preprocessed Data

```python
# Only consider numerical columns for Isolation Forest
features = []
for feature_column in df.select_dtypes(include="number").columns:
    if feature_column != "loan_status":
        features.append(feature_column)

X = df[features].values

isolation_forest_model = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
isolation_forest_model.fit(X)

y_predictions = isolation_forest_model.predict(X)
anomaly_scores = isolation_forest_model.decision_function(X)

df["outlier_prediction"] = y_predictions
df["anomaly_score"] = anomaly_scores

print(f"Total samples for preprocessed data: {len(y_predictions)}")
print(f"Number of outliers for preprocessed data: {(y_predictions == -1).sum()}")
```

# Anomaly Score Distribution of Preprocessed Data

```python
plt.figure(figsize=(8, 5))
plt.hist(anomaly_scores, bins=50, color="blue")
# outlier is less than 0, normal is greater than 0
plt.axvline(0, color="red", linestyle="-", label="Decision boundary")
plt.title("Anomaly Score Distribution")
plt.xlabel("Anomaly Score")
plt.ylabel("Number of samples")
plt.legend()
plt.tight_layout()
plt.show()
```

# Splitting the Preprocessed Dataset into Outliers Only and Normal Datasets

```python
outlier_df = df[df["outlier_prediction"] == -1].drop(columns=["outlier_prediction", "anomaly_score"]).reset_index(drop=True)
normal_df = df[df["outlier_prediction"] == 1].drop(columns=["outlier_prediction", "anomaly_score"]).reset_index(drop=True)

print(f"Outlier dataset sizes {outlier_df.shape}")
print(f"Normal dataset sizes {normal_df.shape}")
```

# Recalculating Anomaly Score Distributions after removing the 5% outliers from both Preprocessed NN and XGB Data

```python
X = normal_df[features].values
isolation_forest_model.fit(X)
anomaly_scores = isolation_forest_model.decision_function(X)

plt.figure(figsize=(8, 5))
plt.hist(anomaly_scores, bins=50, color="blue")
# outlier is less than 0, normal is greater than 0
plt.title("Recalculated Anomaly Score Distribution on Preprocessed Data")
plt.xlabel("Recalculated Anomaly Score")
plt.ylabel("Number of samples")
plt.tight_layout()
plt.show()
```

# Save splitted preprocessed CSV files

```python
outlier_output_path = OUTPUT_DIR / "loan_data_preprocessed_outlier.csv"
normal_output_path = OUTPUT_DIR / "loan_data_preprocessed_normal.csv"

outlier_df.to_csv(outlier_output_path, index=False)
normal_df.to_csv(normal_output_path, index=False)

print("Saved outlier data to:", outlier_output_path)
print("Saved normal data to:", normal_output_path)
```
