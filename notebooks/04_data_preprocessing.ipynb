# Load data

```python
import pandas as pd
from pathlib import Path

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer


# File name
# RAW_DATA_FILE_NAME = "loan_data.csv"
RAW_DATA_FILE_NAME = "loan_data_hard_filtered.csv"

PROJECT_ROOT = Path("..")
RAW_DATA_FILE_PATH = PROJECT_ROOT / "data" / "pre-processed" / RAW_DATA_FILE_NAME

OUTPUT_DIR = PROJECT_ROOT / "data" / "pre-processed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("Raw data path:", RAW_DATA_FILE_PATH)
print("Output folder:", OUTPUT_DIR)

df = pd.read_csv(RAW_DATA_FILE_PATH)
print(df.shape)
df.head()
```

# Data Preprocessing

```python
# Remove Bias column
df = df.drop(columns=['previous_loan_defaults_on_file'])

# Ordinal Mapping
EDUCATION_MAPPING_RULES = {
    'High School': 1,
    'Associate': 2,
    'Bachelor': 3,
    'Master': 4,
    'Doctorate': 5
}

df['person_education'] = df['person_education'].map(EDUCATION_MAPPING_RULES)


print("New DataFrame preprocessed shape:", df.shape)
df.head()
```

## Save preprocessed CSV files

```python
output_path = OUTPUT_DIR / "loan_data_preprocessed.csv"

df.to_csv(output_path, index=False)

print("Saved Pre-processed data to:", output_path)
```
